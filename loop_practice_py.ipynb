```python
text = "the market moved because the market reacted to the news"
words = text.split()

word_count = {}

for word in words:
    if word in word_count:
         word_count[word] += 1
    else:
         word_count[word]  = 1

print(word_count)
```

```python
sentences = [
    "the price of oil rose today",
    "oil prices often rise in winter",
    "today the market reacted to oil news"
]
word_count = {}
for sentence in sentences:
    words = sentence.split()
    for word in words:
       if word in word_count:
          word_count[word] += 1
       else:
          word_count[word] = 1

print(word_count)


```
