```python
raw_records = [
    ["C1", "  jordan lee  ", 120, 80],
    ["C2", "TAYLOR KIM", 200, 150],
    ["C3", "  Riley Chen", 60, 90],
    ["C4", "morgan patel  ", 300, 100],
    ["C5", "  ALEX OWEN  ", 45, 55],
]
def clean_customer(records):
    result = {}
    for row in records:
        customer_id = row[0]
        raw_name = row[1]
        purchase_1 = row[2]
        purchase_2 = row[3]

        clean_name = raw_name.strip().lower()
        average_purchase = (purchase_1 + purchase_2) / 2
        result[customer_id] = { "name": clean_name, "average_purchase": average_purchase
                               }
    return result

cleaned_records = clean_customer(raw_records)
print(cleaned_records)
```

```python
SAMPLE_TICKETS = [
    {"agent": "Agent_A", "resolution_minutes": 30},
    {"agent": "Agent_B", "resolution_minutes": 45},
    {"agent": "Agent_A", "resolution_minutes": -5},
    {"agent": "Agent_B", "resolution_minutes": 20},
    {"agent": "Agent_A", "resolution_minutes": 15},
    {"agent": "Agent_C", "resolution_minutes": 0},
]

def build_report(tickets):
    totals = {}
    for ticket in tickets:
        minutes = ticket["resolution_minutes"]
        if minutes <= 0:
            continue
        agent = ticket["agent"]
        if agent not in totals:
            totals[agent] = minutes
        else:
            totals[agent] += minutes


    return totals

if __name__ == "__main__":
    print(build_report(SAMPLE_TICKETS))
```
