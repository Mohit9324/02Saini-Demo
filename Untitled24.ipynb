```python
import pandas as pd
```

```python
students = pd.read_excel("students.xlsx")
print(students)
```

```python
print(students.loc[[2,3],["name", "city"]])
```

```python
print(students.loc[ :, ["name", "city"]])
```

```python
print(students.loc[ ::2, ["name", "city"]])
```

```python
mask = students["math_score"] > 80
print(students.loc[mask])
```

```python
mask = students["math_score"] > 70
df = students.loc[mask]
print(students.loc[mask])
print(df.loc[2, ["name", "city"]])
```

```python
mask = (students["math_score"] > 70) & (students["city"] == "Chennai")
print(students.loc[mask])
```

```python
mask = (students["math_score"] > 70) | (students["city"] == "Chennai")
print(students.loc[mask])
```
