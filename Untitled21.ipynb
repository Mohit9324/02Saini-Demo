```python
import numpy as np
score = np.array([78, 92, 45, 23, 98, 47])
print(score)
print(score.dtype)
print(score.shape)
```

```python
scores = np.array([78, 92, 55, 23, 88])
print(scores + 5)
print(scores * 2)
print(scores / 100)
print(scores // 10)


print(scores[1: 4])
print(scores[1::2])
print(scores[::-1])
```

```python
scores = np.array([45, 86, 77, 34, 93, 67])
print(scores > 75)
```

```python
scores = np.array([45, 86, 77, 34, 93, 67])
mask = scores > 75
print(mask)
print(scores[mask])
```

```python
sales = np.array([
    [100, 150, 200],
    [300, 250, 400]
])
monthly_bonus = np.array([10, 20, 30])
total_sales = sales + monthly_bonus
print(total_sales)
print(sales.shape)
print(monthly_bonus.shape)
print(sales.shape + monthly_bonus.shape)
```
